```python
"""We kept running into an issue where the notebook could not reliably fetch
MWGC dot dat, even though I could open the link manually.

So instead of, like, working with that, I downloaded the Harris catalogue
as a text file here, and this notebook extracts the relevant Harris part 2
data fields and puts them in a CSV. This is what this file is all about.

It is a preprocessing file designed to make working with the data as smooth
as possible. It is part of the data preparation, not the scientific analysis
itself. It is something I made to help myself."""

from pathlib import Path
import io
import pandas as pd

# we read the harris catalogue
source = Path("mwgc.dat.txt")

lines = source.read_text(encoding="utf-8").splitlines()

# now from what I saw lines 253-409 are the 157 Part II clusters
# python starts counting at zero so we only take 252-409

part2_text = "\n".join(lines[252:409])

# now this si the width of each of the columns in the original text file
widths = [12,7,3,7,6,6,6,8,6,6,6,6,6,15]

columns  = [
    "ID", "[Fe/H]", "wt", "E(B-V)", "V_HB", "(m-M)V", "V_t",
    "M_V,t", "U-B", "B-V", "V-R", "V-I", "spt", "ellip"
]

# now we read Part II without shitfing values whenevr a field is blank

part2 = pd.read_fwf(
    io.StringIO(part2_text),
    widths=widths,
    names=columns
)

# now we basically check the result before saving it.
print("rows and columns: ", part2.shape)
print(part2[part2["ID"] == "Eridanus"])
print(part2[part2["ID"] == "Pal 1"])

# now we save the table for adiitis notebook
part2.to_csv("HarrisPartII.csv", index=False)
```
