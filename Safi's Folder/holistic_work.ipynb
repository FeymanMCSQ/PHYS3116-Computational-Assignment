```python
import pandas as pd

# just like import and then find the ones that are special, classify what part is special
# then print
movement = pd.read_csv("movement_checked.csv")
age = pd.read_csv("age_checked.csv")

movement_ids = set(movement["ID"])
movement_flagged = set(movement.loc[movement["movement_flag"], "ID"])

age_usable = set(age.loc[age["result"] != "not enough to compare", "ID"])

age_flagged = set(age.loc[age["result"].isin(["flagged younger", "flagged older"]), "ID"])

both = movement_flagged & age_flagged

print("The IDs flagged in both were: ", sorted(both))

age_only = []
age_no_movement = []

for cluster_id in age_flagged:
    if cluster_id not in movement_ids:
        age_no_movement.append(cluster_id)
    elif cluster_id not in movement_flagged:
        age_only.append(cluster_id)

movement_only = []
movement_no_usable_age = []

for cluster_id in movement_flagged:
    if cluster_id not in age_usable:
        movement_no_usable_age.append(cluster_id)
    elif cluster_id not in age_flagged:
        movement_only.append(cluster_id)

print("Age was flagged while movement measured but not flagged:", sorted(age_only))
print("Age was flagged while movement unavailable:", sorted(age_no_movement))
print("Movement was flagged while age checked but not flagged:", sorted(movement_only))
print("Movement was flagged while no usable age comparison was found:", sorted(movement_no_usable_age))
```
