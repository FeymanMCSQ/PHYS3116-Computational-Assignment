```python
import pandas as pd

# we load the results from the movement and the age analysis
# these files contain the checked clusters and their flags so I can compare them
# we use the two analysis without repeating their earlier calculation.
movement = pd.read_csv("movement_checked.csv")
age = pd.read_csv("age_checked.csv")

# we keep two movmeent lists: everyone with a usable measurment and the smaller 
# group that passed my movement flag rule. Having both tells me 
# "measured but not flagged" from "not measured"
movement_ids = set(movement["ID"])
movement_flagged = set(movement.loc[movement["movement_flag"], "ID"])

# a usable age result means there were enough similar-metallicity clusters
# so that we could apply the age rule in at least one source
# "not enough to compare" is once again different from "not flagged"
age_usable = set(age.loc[age["result"] != "not enough to compare", "ID"])

# now we collect clusters flagged as either older or younger in either age source
# the list says a source flagged the cluster, it does not mean both sources agreed
age_flagged = set(age.loc[age["result"].isin(["flagged younger", "flagged older"]), "ID"])

# find IDs appearing in both flagged lists
# this is strongest overlap betwee the two screening rules
# but an overlap alone does not really prove a cluster origin.
both = movement_flagged & age_flagged

print("The IDs flagged in both were: ", sorted(both))

# now we split the remining age into two cases, movmvent was measured but did not 
# pass flag rule or no movement to check
age_only = []
age_no_movement = []

for cluster_id in age_flagged:
    if cluster_id not in movement_ids:
        age_no_movement.append(cluster_id)
    elif cluster_id not in movement_flagged:
        age_only.append(cluster_id)

# split the remiaining movement flags into two cases:
# age was checked but did ot pass its flag rules, or 
# there was no usable age comparison
movement_only = []
movement_no_usable_age = []

for cluster_id in movement_flagged:
    if cluster_id not in age_usable:
        movement_no_usable_age.append(cluster_id)
    elif cluster_id not in age_flagged:
        movement_only.append(cluster_id)

# now we print each case seperately so a missing comparison is never treated
# as evidence that the cluster is ordinary
# sort the IDs just to make the lists easier to read and not to rank the clusters.
print("Age was flagged while movement measured but not flagged:", sorted(age_only))
print("Age was flagged while movement unavailable:", sorted(age_no_movement))
print("Movement was flagged while age checked but not flagged:", sorted(movement_only))
print("Movement was flagged while no usable age comparison was found:", sorted(movement_no_usable_age))
```

## Comparing the age and movement results

I matched the clusters by their IDs in both datasets. NGC 1851 was the only cluster flagged in both. Its measured movement is among the highest in the Harris sample, and its age stands out in the Krause data. But van den Berg gives it a different age, so the age evidence for this cluster isn't settled.

Pal 12 and NGC 6171 were both flagged for age, but not for movement. Pal 12 has a similar young age in both age datasets, while NGC 6171's age flag changes when we adjust the rule. Not having a movement flag does not rule out either cluster. It just means their measured speeds did not meet the cutoff I explained earlier.

Seven movement-flagged clusters had a usable age comparison but were not flagged for age. Another seven had no usable age comparison, so I can't say whether their ages are unusual under my rule. All of the age-flagged clusters did have a movement measurement.

Overall, NGC 1851 is the clearest overlap to look into, but even that does not establish that it came from another galaxy. The other groups are useful too, especially where an age measurement is missing or there aren't enough similar clusters to compare.
